**Why did `x * 2` give a different answer the second time?**

At first `x` was 10, so `x * 2` gave 20. When I changed the cell to `x = 50` but did not run it, the kernel still remembered `x = 10`, so `x * 2` still gave 20. After I ran `x = 50`, the kernel stored the new value, so `x * 2` gave 100.

This shows that the kernel remembers the last value that was **run**, not what is written in the cell, so the order in which cells are run matters.

### Study Habits Survey Dataset

This dataset contains survey answers from **80 students** about their study habits. It is simulated data made for teaching, so it does not describe real people.

**Columns in the dataset:**
- `student_id`: unique ID of each student
- `faculty`: the student's faculty
- `year`: year of study (1 to 4)
- `age`: age in years
- `study_hours_per_week`: hours of study per week
- `sleep_hours`: average hours of sleep per night
- `commute_minutes`: minutes travelling to campus each day
- `phone_hours_per_day`: hours on the phone each day
- `gpa`: grade point average (0 to 4)

| Column | Meaning | Type |
|---|---|---|
| faculty | Science, Business, Education, Engineering or Health | text |
| study_hours_per_week | Hours of study per week | number |
| gpa | Grade point average (0 to 4) | number |

Tool used: Python with [pandas](https://pandas.pydata.org/docs/).
